```python
from qiskit import QuantumCircuit
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create circuit
qc = QuantumCircuit(2)

# Gate dropdown
gate_dropdown = widgets.Dropdown(
    options=["X", "Y", "Z", "H", "S", "T"],
    description="Gate:"
)

# Qubit dropdown
qubit_dropdown = widgets.Dropdown(
    options=[0, 1],
    description="Qubit:"
)

# Button
apply_button = widgets.Button(
    description="Apply Gate"
)

# Output area
output = widgets.Output()


def apply_gate(b):
    gate = gate_dropdown.value
    qubit = qubit_dropdown.value

    if gate == "X":
        qc.x(qubit)
    elif gate == "Y":
        qc.y(qubit)
    elif gate == "Z":
        qc.z(qubit)
    elif gate == "H":
        qc.h(qubit)
    elif gate == "S":
        qc.s(qubit)
    elif gate == "T":
        qc.t(qubit)

    with output:
        clear_output(wait=True)
        display(qc.draw("mpl"))


apply_button.on_click(apply_gate)

display(gate_dropdown)
display(qubit_dropdown)
display(apply_button)
display(output)
```
